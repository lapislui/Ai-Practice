```python
import AI_data as aid

loader = aid.LoadData("data/student_ml_linked.csv")
df = loader.read()

df.info()
print("Head:" , df.head())
print("\ncollumns",df.columns)
print("\nNull Vallues:",df.isna().sum())
print("\nDescribe:",df.describe())
```

```python
print(df["study_hours_per_day"].mean())
print(df["study_hours_per_day"].median())
print(df["study_hours_per_day"].std())
```

```python
import matplotlib.pyplot as plt

plt.scatter(df["study_hours_per_day"], df["final_exam_score"])
plt.xlabel("Study hours per day")
plt.ylabel("Final exam score")
plt.title("Study time vs final exam score")
plt.show()
```

```python
import AI_data as aid

loader = aid.LoadData("data/student_ml_linked.csv")
df = loader.read()

reg = aid.SimpleLinearRegression(df)
reg.fit("study_hours_per_day", "final_exam_score")
reg.summary()

print(reg.predict(5))
reg.plot()
```
